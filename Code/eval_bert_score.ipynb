```python
from bert_score import score

def evaluate_bertscore(references, candidates, lang='sl'):
    P, R, F1 = score(candidates, references, lang=lang, rescale_with_baseline=True)
    avg_f1 = F1.mean().item()
    return F1.tolist(), avg_f1, P, R

def eval_responses(responses):
    references = responses['completion'].tolist()
    candidates = responses['output'].tolist()
    return evaluate_bertscore(references, candidates)
```

```python
import pandas as pd
results_generated = pd.read_csv("eval_results_generated_reports.csv")
results_generated = results_generated.dropna()
results_original = pd.read_csv("eval_results_original_reports.csv")
results_original = results_original.dropna()
results_gemini = pd.read_csv("eval_results_gemini.csv")
results_gemini = results_gemini.dropna()
results_base_gams = pd.read_csv("eval_results_base_gams.csv")
results_base_gams = results_base_gams.dropna()
```

```python
bertscore_f1_generated, avg_bertscore_generated, P_generated, R_generated = eval_responses(results_generated)
bertscore_f1_original, avg_bertscore_original, P_original, R_original = eval_responses(results_original)
bertscore_f1_gemini, avg_bertscore_gemini, P_gemini, R_gemini = eval_responses(results_gemini)
bertscore_f1_base_gams, avg_bertscore_base_gams, P_base_gams, R_base_gams = eval_responses(results_base_gams)
print(f"BERTScore for generated responses F1: {avg_bertscore_generated:.4f} P: {P_generated.mean().item():.4f} R: {R_generated.mean().item():.4f}")
print(f"BERTScore for original responses F1: {avg_bertscore_original:.4f} P: {P_original.mean().item():.4f} R: {R_original.mean().item():.4f}")
print(f"BERTScore for Gemini responses F1: {avg_bertscore_gemini:.4f} P: {P_gemini.mean().item():.4f} R: {R_gemini.mean().item():.4f}")
print(f"BERTScore for Base GAMS responses F1: {avg_bertscore_base_gams:.4f} P: {P_base_gams.mean().item():.4f} R: {R_base_gams.mean().item():.4f}")
```
